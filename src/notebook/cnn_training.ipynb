# CNN Training (Shared / Non-Shared)
Notebook modular untuk training CNN berbasis konfigurasi eksperimen.

```python
from pathlib import Path

from src.cnn.experiments import all_shared_experiments, get_experiment
from src.cnn.train_cnn import train_one

data_root = Path('src/dataset')
output_dir = Path('outputs/cnn')
image_size = 64
experiment_id = 'd1_f16_k3_max'
non_shared = False
epochs = 1
batch_size = 16

output_dir.mkdir(parents=True, exist_ok=True)
```

## Train satu eksperimen

```python
if (data_root / 'train').exists() and (data_root / 'val').exists() and (data_root / 'test').exists():
    config = get_experiment(experiment_id, image_size=image_size)
    exp_dir = train_one(
        config=config,
        data_root=data_root,
        output_dir=output_dir,
        epochs=epochs,
        batch_size=batch_size,
        non_shared=non_shared,
    )
    exp_dir
else:
    print('Dataset belum tersedia. Silakan set data_root terlebih dulu.')
```

## Train semua shared experiments (butuh waktu lama)

```python
run_all = False  # ganti ke True jika siap training penuh

if run_all and (data_root / 'train').exists():
    for config in all_shared_experiments(image_size=image_size):
        train_one(
            config=config,
            data_root=data_root,
            output_dir=output_dir,
            epochs=epochs,
            batch_size=batch_size,
            non_shared=False,
        )
    print('Selesai run_all shared experiments')
else:
    print('run_all=False atau dataset belum tersedia.')
```
